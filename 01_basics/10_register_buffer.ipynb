```python
import torch
from torch import nn, tensor
```

### Simple

```python
class MyNetwork(nn.Module):
    def __init__(self):
        super().__init__()

        self.register_buffer(
            "scale",
            torch.tensor(0.5)
        )

    def forward(self, x):
        return x * self.scale
```

```python
model = MyNetwork()

print(model.scale)
```

### Include in State Dict

```python
class MyNetwork(nn.Module):
    def __init__(self):
        super().__init__()

        self.linear1 = nn.Linear(4, 8)
        self.linear2 = nn.Linear(8, 2)

        self.register_buffer(
            "scale1",
            torch.tensor(0.5)
        )

        self.register_buffer(
            "scale2",
            torch.tensor(2.0)
        )

    def forward(self, x):
        x = self.linear1(x)
        x = x * self.scale1

        x = self.linear2(x)
        x = x * self.scale2

        return x
```

```python
model = MyNetwork()

print(model)
print(model.state_dict().keys())
```

### Not using register buffer

```python
class MyNetwork(nn.Module):
    def __init__(self):
        super().__init__()

        self.linear1 = nn.Linear(4, 8)
        self.linear2 = nn.Linear(8, 2)

        self.scale1 = tensor(0.5)
        self.scale2 = tensor(0.2)

    def forward(self, x):
        x = self.linear1(x)
        x = x * self.scale1

        x = self.linear2(x)
        x = x * self.scale2

        return x
```

```python
model = MyNetwork()

print(model)
print(model.state_dict().keys())
```
