```python
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self, n_layers=100):
        super().__init__()

        layers = []

        for _ in range(n_layers - 1):
            layers.append(nn.Linear(1000, 1000))

        layers.append(nn.Linear(1000, 1))

        self.layers = nn.Sequential(*layers)

    def forward(self, x):
        return self.layers(x)
```

```python
big_model = MyModel(n_layers=100)
```

```python
y_pred = big_model(x)
```

```python
y_pred.shape
```

```python
loss_fn = nn.MSELoss()
```

```python
loss_fn(y_pred, y_true)
```

### Batching

```python
batch_size = 100
losses = []
model = nn.Linear(1000,1)

for i in range(0, len(x), batch_size):
    x_batch = x[i:i + batch_size]
    y_batch = y_true[i:i + batch_size]

    y_pred = model(x_batch)
    loss = loss_fn(y_pred, y_batch)

    losses.append(loss)
```

```python
loss
```

### Epoch

```python
epochs = 10
batch_size = 100
model = nn.Linear(1000,1)

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.001
)

for _ in range(epochs):
    for i in range(0, len(x), batch_size):
        x_batch = x[i:i + batch_size]
        y_batch = y_true[i:i + batch_size]
    
        y_pred = model(x_batch)
        loss = loss_fn(y_pred, y_batch) 
        
        loss.backward()
        optimizer.step()

        optimizer.zero_grad()
```
