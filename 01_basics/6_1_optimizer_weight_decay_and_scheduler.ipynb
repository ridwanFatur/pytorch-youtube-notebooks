```python
from torch import tensor, nn
import torch
```

### Weight Decay

```python
model = nn.Linear(1,1, bias = False)
model.load_state_dict({
    "weight": torch.tensor([[1]]),
})

loss_fn = nn.MSELoss()
x = tensor([10.0])
y_true = tensor([15.0])

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.005,
    # weight_decay=0.001
)
```

```python
y_pred = model(x)
loss = loss_fn(y_pred, y_true)
print("Loss:", loss)
loss.backward()
print("Model Grad:", model.weight.grad)
optimizer.step()
print("Model Weight:", model.weight.item())
optimizer.zero_grad()
```

### Non Weight Decay

```python
model = nn.Linear(1,1, bias = False)
model.load_state_dict({
    "weight": torch.tensor([[1]]),
})

loss_fn = nn.MSELoss()
x = tensor([10.0])
y_true = tensor([15.0])

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.005,
    weight_decay=0.001
)
```

```python
y_pred = model(x)
loss = loss_fn(y_pred, y_true)
print("Loss:", loss)
loss.backward()
print("Model Grad:", model.weight.grad)
optimizer.step()
print("Model Weight:", model.weight.item())
optimizer.zero_grad()
```

### Learning Rate Changing

```python
model = nn.Linear(1,1, bias = False)
model.load_state_dict({
    "weight": torch.tensor([[1]]),
})

loss_fn = nn.MSELoss()
x = tensor([10.0])
y_true = tensor([15.0])

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.005,
)

scheduler = torch.optim.lr_scheduler.StepLR(
    optimizer,
    step_size=1,
    gamma=0.9
)
```

```python
for _ in range(10):
    optimizer.step()
    scheduler.step()
    print("LR:", optimizer.param_groups[0]["lr"])
```
