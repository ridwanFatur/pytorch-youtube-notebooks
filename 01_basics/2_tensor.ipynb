### Gradient

```python
x = tensor([[2, 3], [2, 3], [2, 3]])
```

```python
x.grad
```

```python
x.requires_grad
```

```python
_input = tensor(4.0)
_param = tensor(5.0)
_param.requires_grad = True
```

```python
y_pred = _input * _param
y_pred
```

```python
y_target = tensor(25.0)
```

```python
loss = y_target - y_pred
loss
```

```python
_param.grad
```

```python
loss.backward()
```

```python
_param.grad
```

```python
_input = tensor(4.0)
_param = tensor(5.0)
_param.requires_grad = True

with torch.no_grad():
    y_pred = _input * _param
    y_target = tensor(25.0)
    loss = y_target - y_pred
```

```python
loss
```

```python
loss.backward()
```
