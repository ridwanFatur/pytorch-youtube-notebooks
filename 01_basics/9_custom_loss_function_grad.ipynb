```python
import torch
from torch import nn, tensor
```

```python
lin = nn.Linear(2, 1)

lin.load_state_dict({
    "weight": torch.tensor([[0.1, 0.3]]),
    "bias": torch.tensor([-0.5]),
})
```

```python
x = torch.tensor([[1.0, 2.0]])
y = lin(x)
```

```python
y
```

```python
def custom_loss(y):
    if y > 0:
        loss = torch.abs(y**3 - 5)
    else:
        loss = torch.abs(y**0.5 - 2)
    return loss
```

```python
loss = custom_loss(y)
loss
```

```python
loss.backward()
```

```python
lin.weight.grad, lin.bias.grad
```

```python
dL_dy = - 3 * (y.item()) ** 2
dy_dweight1 = x[0][0]
dy_dweight2 = x[0][1]
dy_dbias = 1

dL_dweight1 = dL_dy * dy_dweight1
dL_dweight2 = dL_dy * dy_dweight2
dL_dbias = dL_dy * dy_dbias
```
