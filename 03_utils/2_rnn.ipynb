```python
### Seq Length > 1
rnn = nn.RNN(
    input_size=1,
    hidden_size=1,
    bias=False,
    batch_first=True,
)
state_dict = {
    "weight_ih_l0": torch.tensor([
        [3.]
    ]),
    "weight_hh_l0": torch.tensor([
        [4.]
    ])
}
rnn.load_state_dict(state_dict)

x = torch.tensor([
    [[0.1], [0.2]],
])

output, hidden = rnn(x)
output, hidden
```

```python
W_ih = rnn.weight_ih_l0.item()
W_hh = rnn.weight_hh_l0.item()

print(W_ih)
print(W_hh)
```

```python
h0 = 0

x1 = 0.1
x2 = 0.2
```

```python
z1 = torch.tensor(x1 * W_ih + h0 * W_hh)
z1
```

```python
h1 = torch.tanh(z1)
h1
```

```python
z2 = torch.tensor(x2 * W_ih + h1.item() * W_hh)
z2
```

```python
h2 = torch.tanh(z2)
h2
```
