```python
from torch import tensor, nn
import torch
```

### 1 Layer

```python
model = nn.Linear(1,1, bias = False)
model.load_state_dict({
    "weight": torch.tensor([[1]]),
})
```

```python
model.weight
```

```python
loss_fn = nn.MSELoss()
```

```python
x = tensor([10.0])
y_true = tensor([15.0])
y_pred = model(x)
y_pred
```

```python
loss = loss_fn(y_pred, y_true)
loss
```

```python
loss.backward()
```

```python
model.weight.grad
```

```python
# grad = dl/dw -> dL_dw
```

```python
2 * (10 - 15) * 10
```

### > 1 layer

```python
model = nn.Sequential(
    nn.Linear(1, 1, bias = False),
    nn.Linear(1, 1, bias = False),
)

model.load_state_dict({
    "0.weight": torch.tensor([[2.0]]),
    "1.weight": torch.tensor([[3.0]]),
})
```

```python
model[0].weight, model[1].weight
```

```python
x = tensor([7.0])
y_true = tensor([15.0])
y_pred = model(x)
loss = loss_fn(y_pred, y_true)
loss.backward()
```

```python
y_pred
```

```python
model[0].weight.grad, model[1].weight.grad
```
