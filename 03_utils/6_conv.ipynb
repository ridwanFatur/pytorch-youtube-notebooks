### Grad

```python
x = torch.tensor([
    [
        [1., 2., 3., 4., 5., 6.]
    ]
])
conv = nn.Conv1d(
    in_channels=1,
    out_channels=1,
    kernel_size=2,
    bias=False,
    stride=2
)

state_dict = {
    "weight": torch.tensor([
        [
            [2., 3.]
        ]
    ])
}
conv.load_state_dict(state_dict)
```

```python
y_pred = conv(x)
y_pred
```

```python
y_target = tensor([[[ 10., 15., 30.]]])
```

```python
loss_fn = nn.MSELoss()

loss = loss_fn(y_pred, y_target)
loss
```

```python
loss.backward()
```

```python
conv.weight.grad
```

```python
w0 = tensor(2.0)
w1 = tensor(3.0)

x0 = tensor(1.)
x1 = tensor(2.)
x2 = tensor(3.)
x3 = tensor(4.)
x4 = tensor(5.)
x5 = tensor(6.)

y_pred0 = x0 * w0 + x1 * w1
y_pred1 = x2 * w0 + x3 * w1
y_pred2 = x4 * w0 + x5 * w1

y_target0 = tensor(10.)
y_target1 = tensor(15.)
y_target2 = tensor(30.)

L0 = (y_pred0 - y_target0) ** 2
L1 = (y_pred1 - y_target1) ** 2
L2 = (y_pred2 - y_target2) ** 2

L = (L0 + L1 + L2) / 3

dL0_dypred0 = 2 * (y_pred0 - y_target0)
dL1_dypred1 = 2 * (y_pred1 - y_target1)
dL2_dypred2 = 2 * (y_pred2 - y_target2)
```

```python
dypred0_dw0 = x0
dypred1_dw0 = x2
dypred2_dw0 = x4

dL0_dw0 = dL0_dypred0 * dypred0_dw0
dL1_dw0 = dL1_dypred1 * dypred1_dw0
dL2_dw0 = dL2_dypred2 * dypred2_dw0

dL_dw0 = (dL0_dw0 + dL1_dw0 + dL2_dw0) / 3
dL_dw0
```
