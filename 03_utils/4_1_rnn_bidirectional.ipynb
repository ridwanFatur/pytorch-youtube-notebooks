```python
import torch
import torch.nn as nn
from torch import tensor
```

```python
# Seq Length = 0
rnn = nn.RNN(
    input_size=1,
    hidden_size=1,
    bias=False,
    bidirectional=True,
    batch_first=True
)

state_dict = {
    # Forward
    "weight_ih_l0": torch.tensor([
        [0.3]
    ]),
    "weight_hh_l0": torch.tensor([
        [0.4]
    ]),

    # Backward
    "weight_ih_l0_reverse": torch.tensor([
        [0.5]
    ]),
    "weight_hh_l0_reverse": torch.tensor([
        [0.6]
    ])
}
rnn.load_state_dict(state_dict)
```

```python
x = torch.tensor([
    [
        [0.1],
    ]
])
x.shape
```

```python
output, hidden = rnn(x)
output, hidden
```

```python
output.shape, hidden.shape
```

```python
W_ih = rnn.weight_ih_l0.item()
W_hh = rnn.weight_hh_l0.item()

h0 = 0
x1 = 0.1

z1 = torch.tensor(x1 * W_ih + h0 * W_hh)
h1 = torch.tanh(z1)
h1
```

```python
W_ih_reverse = rnn.weight_ih_l0_reverse.item()
W_hh_reverse = rnn.weight_hh_l0_reverse.item()

h0_reverse = 0

x1 = 0.1

z1_reverse = torch.tensor(x1 * W_ih_reverse + h0_reverse * W_hh_reverse)
h1_reverse = torch.tanh(z1_reverse)
h1_reverse
```

```python
# Seq Length > 1
x = torch.tensor([
    [
        [0.1], [0.2], [0.3],
    ]
])
x.shape
```

```python
output, hidden = rnn(x)
output, hidden
```

```python
output.shape, hidden.shape
```

```python
W_ih = rnn.weight_ih_l0.item()
W_hh = rnn.weight_hh_l0.item()

h0 = 0
x1 = 0.1
x2 = 0.2
x3 = 0.3

z1 = torch.tensor(x1 * W_ih + h0 * W_hh)
h1 = torch.tanh(z1)

z2 = torch.tensor(x2 * W_ih + h1.item() * W_hh)
h2 = torch.tanh(z2)

z3 = torch.tensor(x3 * W_ih + h2.item() * W_hh)
h3 = torch.tanh(z3)

print(h1, h2, h3)
```

```python
W_ih_reverse = rnn.weight_ih_l0_reverse.item()
W_hh_reverse = rnn.weight_hh_l0_reverse.item()

h0_reverse = 0

x1 = 0.1
x2 = 0.2
x3 = 0.3

z3_reverse = torch.tensor(
    x3 * W_ih_reverse + h0_reverse * W_hh_reverse
)
h3_reverse = torch.tanh(z3_reverse)

z2_reverse = torch.tensor(
    x2 * W_ih_reverse + h3_reverse.item() * W_hh_reverse
)
h2_reverse = torch.tanh(z2_reverse)

z1_reverse = torch.tensor(
    x1 * W_ih_reverse + h2_reverse.item() * W_hh_reverse
)
h1_reverse = torch.tanh(z1_reverse)

print(h1_reverse, h2_reverse, h3_reverse)
```
