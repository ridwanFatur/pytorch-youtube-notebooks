### Grad

```python
model = nn.Sequential(
    nn.Linear(1, 2, bias=False),
    nn.ReLU(),
    nn.Linear(2, 1, bias=False)
)

state_dict = {
    "0.weight": torch.tensor([
        [-1.],
        [2.]
    ]),
    
    "2.weight": torch.tensor([
        [3., 4.]
    ])
}

model.load_state_dict(state_dict)

x = tensor([1.])
y = model(x)
print("y:", y)

loss_fn = nn.MSELoss()
target = tensor([10.])

loss = loss_fn(y, target)
print("loss:", loss)

loss.backward()
```
