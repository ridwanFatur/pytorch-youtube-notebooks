### ML Flow - Sqlite Tracking Uri

```python
mlflow.set_tracking_uri("sqlite:///test.db")
```

```python
mlflow.set_experiment("Simple-Experiment")
```

```python
%%bash
cat <<EOF > artifact.txt
Model-Testing-Artifact
EOF
```

```python
%%bash
mkdir -p artifact

cat <<EOF > artifact/artifact-1.txt
Model-Testing-Artifact-1
EOF

cat <<EOF > artifact/artifact-2.txt
Model-Testing-Artifact-2
EOF

cat <<EOF > artifact/artifact-3.txt
Model-Testing-Artifact-3
EOF
```

```python
timestamp = datetime.now().strftime("%Y%m%d-%H%M%S")
custom_run_name = f"Experiment_{timestamp}"
print(custom_run_name)

with mlflow.start_run(run_name=custom_run_name):
    mlflow.log_param("param_1", 1)
    mlflow.log_params({"param_key_1":1})

    for step in range(11):
        value = 0.1 + (0.7 - 0.1) * step / 10
        mlflow.log_metric("metric_1", value, step=step)
    mlflow.log_metrics({"metric_key":1})

    mlflow.log_artifact("artifact.txt")
    mlflow.log_artifacts("artifact", artifact_path="artifact_folder")

    mlflow.set_tag("tag", "tag-1")
    mlflow.set_tags({"tag-key":"tag-key-1"})

    mlflow.log_text("Log Text", "log_text.txt")
    mlflow.log_dict({"key":"value"}, "log_dict.json")
```
